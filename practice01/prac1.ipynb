```python
temp = int(input("현재 온도를 입력해주세요: "))

if 25 < temp < 40:
    print("여름")
elif 10 <= temp <= 25:
    print("가을, 봄")
elif temp < 10:
    print("겨울")
elif temp >= 40:
    print("한국이 아닌 것 같습니다..")
```

```python
num = 0
count = 0

while True:
    if num < 5:
        num += 1
        count += 1
        print(f"5를 만날때까지 계속 1을 더하자 현재 수: {num}, 횟수 : {count}")
    elif num == 5:
        print(f"드디어 5를 만났다. 횟수 : {count}")
        break

print("5를 만나서 끝")
        
```

```python
# 구구단을 만들어보자

for i in range(2,10,1):
    print(f"----{i}단----")
    for o in range(1,10,1):
        print(f"{i} x {o} = {i*o}")
    if i == 3:
        print(f"{i}단까지만 하자")
        break    
```

```python
user_info = [
    {"name" : "k", "level" : "VIP", "point" : 1100},
    {"name" : "s", "level" : "Platium", "point" : 1000},
    {"name" : "d", "level" : "Sliver","point" : 500},
    {"name" : "f", "level" : "Gold", "point" : 200}
]

user_count = 0

for user in user_info:
    if user["point"] >= 1000:
        print("입장이 가능합니다.")
        user_count += 1
    else:
        print("물건을 구입하여 포인트를 모으시오")
        
print(f"입장 가능한 인원은 {user_count}명입니다.")
```
